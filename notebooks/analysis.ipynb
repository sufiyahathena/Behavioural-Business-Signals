````python
import sys
sys.path.insert(0, '..')  # add project root to path

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

import config
````

````python
# Load scores
fin = pd.read_csv('../data/scores/financial_scores.csv')
beh = pd.read_csv('../data/scores/behavioural_scores.csv')

combined = fin[['ticker', 'company_name', 'financial_score']].merge(
    beh[['ticker', 'behavioural_score']], on='ticker', how='outer'
)
combined.head()
````

````python
# Scatter plot: Financial Score vs Behavioural Score
fig = px.scatter(
    combined,
    x='financial_score',
    y='behavioural_score',
    text='ticker',
    title='Financial vs Behavioural Score',
    labels={
        'financial_score': 'Financial Score (0–100)',
        'behavioural_score': 'Behavioural Score (0–100)',
    },
    width=800,
    height=600,
)

# Quadrant lines
fig.add_hline(y=50, line_dash='dash', line_color='grey', opacity=0.5)
fig.add_vline(x=50, line_dash='dash', line_color='grey', opacity=0.5)

fig.update_traces(textposition='top center', marker_size=10)
fig.show()
````

````python
# Divergence: companies where financial and behavioural scores differ most
combined['divergence'] = (combined['financial_score'] - combined['behavioural_score']).abs()
combined.sort_values('divergence', ascending=False).head(10)
````
